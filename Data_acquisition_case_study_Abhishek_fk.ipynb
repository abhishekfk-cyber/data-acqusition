```python
# Task 6
# Rename columns wherever necessary so that all datasets have a common column
# name for merging.
```

```python
gdp_df.rename(columns={'Country': 'country'}, inplace=True)
gdp_df

```

```python
# Task 7
# Convert the required columns into appropriate numeric data types.
```

```python
# Convert the required columns into appropriate numeric data types.
gdp_df['GDP'] = pd.to_numeric(gdp_df['GDP'])
print(gdp_df.dtypes)
```

```python
lit_rate_df.rename(columns={'@name': 'country'}, inplace=True)
```

```python
# Task 8
# Merge all four datasets into one DataFrame.

```

```python
df_merged = pd.merge(gdp_df, int_users_df, on='country', how='inner')
df_merged = pd.merge(df_merged, lit_rate_df, on='country', how='inner')
df_merged = pd.merge(df_merged, population_df, on='country', how='inner')
df_merged
```

```python
# Task 9
# Create the following calculated columns:
# 1. Internet Penetration (%)
# 2. GDP Per Capita
```

```python
df_merged["Internet Penetration (%)"] = (df_merged["internet_users"] / df_merged["population"]) * 100
df_merged["GDP Per Capita"] = (df_merged["GDP"] / df_merged["population"])
df_merged
```

```python
# Task 10
# Create a SQLite database. Store the final merged dataset into a table named
# ‘country_statistics’


```

```python
import sqlite3
conn = sqlite3.connect('country_statistics.db')
df_merged.to_sql('country_statistics', conn, if_exists='replace', index=False)
conn.close()
print("DataFrame stored in SQLite database successfully.")
```
